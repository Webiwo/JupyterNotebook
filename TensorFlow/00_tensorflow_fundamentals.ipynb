```python
mat1 = tf.constant(np.arange(1, 7, dtype=np.int32), shape=(2, 3))
mat2 = tf.constant(np.arange(7, 13, dtype=np.int32), shape=(2, 3))
mat1, mat2
```

```python
tf.matmul(mat1, mat2, transpose_b=True)
```

```python
mat3 = tf.transpose(mat2)
mat3
```

```python
tf.matmul(mat1, mat3)
```

```python
mat1, tf.reshape(mat1, shape=(3, 2)), tf.transpose(mat1)
```

```python
tf.tensordot(mat1, mat3, axes=1) # the same as tf.matmul(mat1, mat3)
```

### Tensor aggregation

```python
E = tf.constant(np.random.randint(0, 100, size=50))
E
```

```python
tf.reduce_min(E), tf.reduce_max(E), tf.reduce_mean(E), tf.reduce_sum(E)
```

```python
tf.math.reduce_variance(tf.cast(E, dtype=tf.float16))
```

```python
tf.math.reduce_std(tf.cast(E, dtype=tf.float16))
```

### Find the positional min and max

```python
tf.random.set_seed(42)
F = tf.random.uniform(shape=[50])
F
```

```python
tf.argmax(F) # index on our largest value position
```

```python
F[tf.argmax(F)]
```

```python
tf.math.equal(F[tf.argmax(F)], tf.reduce_max(F))
```

```python
tf.size(F)
```

```python
G = tf.reshape(F, shape=(10, 5))
G
```

```python
tf.argmax(G, axis=1)
```

### Squeezing a tnesor (removing all single dimensions)

```python
G_ext = tf.expand_dims(tf.expand_dims(G, axis=0), axis=0)
G_ext
```

```python
G_ext.shape
```

```python
G_zqueezed = tf.squeeze(G_ext)
G_zqueezed
```
